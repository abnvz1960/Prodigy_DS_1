```python
#Task 1 Visualizing student data using Bar chart and histogram
#Objectve - Creating a bar chart and a histogram to visualize the distribution of a categorical 
#variable (Gender) and a continuous variable (Math Score).

#importing libaries 
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

#setting theme
sns.set_theme(style='whitegrid')

#loading dataset
# (Replace 'dataset.csv' with your actual file path if different)
df = pd.read_csv(r"C:\Users\abnvz\Downloads\Student Performance.csv")

#exploring datset
print('First 5 Rows')
print(df.head())

print('\n Dataset Info')
print(df.info())

print('\n Dataset Shape (Rows, Columns)')
print(df.shape)

print('\n Column Names')
print(df.columns.tolist())

#data cleaning
gender_col = 'gender'
score_col = 'math score'

print('\n Missing values in target columns:')
print(df[[gender_col, score_col]].isnull().sum())

#drop rows with missing values
df = df.dropna(subset=[gender_col, score_col])
print(f'Data cleaning complete. Remaining rows: {df.shape[0]}')

#bar chart for categorical variable gender
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x=gender_col, palette='Set2')
plt.title('Distribution of Gender', fontsize=12, pad=10)
plt.xlabel('Gender', fontsize=10)
plt.ylabel('Count', fontsize=10)
plt.show()

#histogram for cont variable math score
plt.figure(figsize=(8, 5))
sns.histplot(
    data=df,
    x=score_col,
    bins=10,
    kde=True,
    color='skyblue',
    edgecolor='black',
)
plt.title('Distribution of Math Scores', fontsize=12, pad=10)
plt.xlabel('Math Score', fontsize=10)
plt.ylabel('Frequency', fontsize=10)
plt.show()
```

```python
#Interpretation
print(" BAR CHART INTERPRETATION")
print("\n The bar chart clearly shows the count of students in each gender category ")
print("\n It helps us quickly see which category has the highest representation and whether the sample is balanced across genders.")
print("\n ------------")
print("\n HISTOGRAM INTERPRETATION")
print("\n The histogram shows how student math scores are spread out across different score ranges.")
print("\n The smooth curve (KDE) helps identify the general shape of the scores, revealing the most common ")
print("score range and whether scores are skewed towards higher or lower values")
```
